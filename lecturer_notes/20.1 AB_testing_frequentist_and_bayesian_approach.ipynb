**Question: What is a hypothesis testing?**

Hypothesis testing is a statistical method used to determine whether a hypothesis or assumption about a population is supported by the available data.

It involves setting up two hypotheses, the null hypothesis (H0) and the alternative hypothesis (Ha) and using sample data to calculate a test statistic and a p-value. The p-value is compared to a chosen significance level (alpha) to determine whether to reject the null hypothesis and accept the alternative hypothesis, or fail to reject the null hypothesis. Hypothesis testing is commonly used in scientific research, quality control, and other fields to make decisions based on available data.

### Null Hypothesis testing
The null hypothesis (H0) is a statement or assumption that there is no significant difference or relationship between two or more variables in a population. In other words, it’s the default position that there is no effect or change. The purpose of hypothesis testing is to gather evidence (data) to either reject or fail to reject the null hypothesis in Favour of the alternative hypothesis, which claims there is a significant effect or relationship.

### Alternate Hypothesis
Alternative hypothesis (H1 or Ha) is a statement or assumption that proposes a significant difference or relationship between two or more variables. It’s the opposite of the null hypothesis and represents an alternative explanation that the observed data is not due to chance or random variation.

It is often used in scientific research, economics, medicine, and various other disciplines. Below are a few examples illustrating how the null hypothesis (H0) and alternative hypothesis (H1 or Ha) are formulated and tested:

### Example 1: Clinical Trial for a New Drug
A pharmaceutical company develops a new drug claimed to lower blood pressure. In this scenario:
- **Null Hypothesis (H0)**: The new drug has no effect on blood pressure.
- **Alternative Hypothesis (H1)**: The new drug lowers blood pressure.
  
The researchers would collect data from a sample of individuals, administering the drug and measuring blood pressure before and after. If the data shows a significant reduction in blood pressure, they might reject the null hypothesis in favor of the alternative.

### Example 2: Educational Intervention
An educational institution implements a new teaching method aimed at improving student performance in mathematics.
- **Null Hypothesis (H0)**: The new teaching method has no effect on student performance.
- **Alternative Hypothesis (H1)**: The new teaching method improves student performance.
  
Data would be collected on student performance before and after the implementation of the new teaching method. If there's a significant improvement in performance, the null hypothesis might be rejected in favor of the alternative.

### Example 3: Business Sales Strategy
A business implements a new sales strategy to increase revenue.
- **Null Hypothesis (H0)**: The new sales strategy has no effect on revenue.
- **Alternative Hypothesis (H1)**: The new sales strategy increases revenue.
  
By comparing sales data before and after the implementation of the new strategy, the business can determine whether the new strategy has a significant effect on revenue.

In these examples, the objective is to gather data to either support or refute the null hypothesis.

**Question: What are the steps involved in Hypothesis Testing?**


1. Formulate a Null and Alternate hypothesis: This step involves defining the null hypothesis (H0) and alternative hypothesis (H1 or Ha). The null hypothesis assumes that there is no significant difference or relationship between two or more variables, while the alternative hypothesis proposes a significant difference or relationship.
2. Select a significance level: This step involves selecting a level of significance (alpha) that represents the probability of rejecting the null hypothesis when it is actually true. The most common levels of significance are 0.05 and 0.01.
3. Check assumptions: This step involves checking if the data satisfies certain assumptions for the statistical test being used. For example, if using a t-test, one needs to check whether the data is normally distributed.
4. Decide which test is appropriate: This step involves selecting the appropriate statistical test to use for the hypothesis test, based on the type of data and research question being investigated. Common tests include Z-test, T-test, Chi-square test, and ANOVA.
5. State the relevant test statistic: This step involves defining the test statistic that will be used to evaluate the hypothesis. The test statistic will vary depending on the type of test being conducted.
6. Conduct the test: This step involves calculating the test statistic and determining the p-value. The p-value represents the probability of observing the test statistic, assuming the null hypothesis is true.
7. Reject or not reject the Null Hypothesis: This step involves comparing the p-value to the level of significance. If the p-value is less than or equal to the level of significance, the null hypothesis is rejected in favor of the alternative hypothesis. If the p-value is greater than the level of significance, the null hypothesis is not rejected.
8. Interpret the result: This step involves interpreting the result of the hypothesis test, based on whether the null hypothesis was rejected or not. The interpretation should be based on the research question and the context of the study.

### Example:

The company claims the loyalty program changed the average monthly purchases per customer to 5. A random sample of 50 customers shows an average of 4.5 purchases with a standard.deviation of 0.5 purchases.

**Question: What will be Null hypothesis and Alternative Hypothesis be?**

   - **Null Hypothesis (H0):** μ = 5 (The population mean monthly purchases per customer is 5)
   - **Alternative Hypothesis (Ha):** μ ≠ 5 (The population mean monthly purchases per customer is not 5)

**Question: Let's use simple two sided z test. what will be the calculation be?**

   - **Z-Score Calculation:**
$$ Z = \frac{(4.5 - 5)}{(0.5 / \sqrt{50})} = -7.07$$


```python
import math
from scipy import stats
import matplotlib.pyplot as plt
import numpy as np

# Given data
sample_mean = 4.5  # average purchases per customer from sample
population_mean = 5  # claimed average purchases per customer
sample_std_dev = 0.5  # standard deviation of purchases per customer from sample
sample_size = 50  # number of customers in the sample

# Calculating the z-score
z_score = (sample_mean - population_mean) / (sample_std_dev / math.sqrt(sample_size))

# Getting the p-value for two-tailed test
p_value = stats.norm.sf(abs(z_score)) * 2  # multiply by 2 for two-tailed test

# Printing the results
print(f'Z-Score: {z_score:.2f}')
print(f'P-Value: {p_value:.2f}')

# Decision making at 0.05 significance level
alpha = 0.05
if p_value < alpha:
    print(f'At {alpha} significance level, we reject the null hypothesis.')
else:
    print(f'At {alpha} significance level, we fail to reject the null hypothesis.')
```

```python
# Adjusting the provided code to depict a two-sided test

plt.figure(figsize=(10, 6))
plt.xlim(4, 6)
plt.ylim(0, 5)
mu = population_mean  # mean
sigma = sample_std_dev / math.sqrt(sample_size)  # standard error
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)
plt.plot(x, stats.norm.pdf(x, mu, sigma))

pdf = stats.norm.pdf(x, mu, sigma)

# Plotting the normal distribution remains the same
plt.plot(x, pdf, label="Normal Distribution (μ = 5)", color='green')
plt.axvline(x=sample_mean, color='red', linestyle='--', linewidth=2, label=f'Sample Mean (x̄ = {sample_mean})')

# Shading the rejection region for two-sided test (both tails)
critical_value_two_sided = stats.norm.ppf(1 - 0.025) * sigma  # critical value for alpha=0.05, two-tailed
x2_lower = np.linspace(4, mu - critical_value_two_sided, 100)
x2_upper = np.linspace(mu + critical_value_two_sided, 6, 100)
plt.fill_between(x2_lower, stats.norm.pdf(x2_lower, mu, sigma), color='lightblue', alpha=0.7, label='Rejection Region (Two-sided)')
plt.fill_between(x2_upper, stats.norm.pdf(x2_upper, mu, sigma), color='lightblue', alpha=0.7)

# Adding labels, title, and legend
plt.title('Two-sided Test for Loyalty Program Effect')
plt.xlabel('Average Monthly Purchases per Customer')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

**Question: How will you make the desicion?**

The Z-score of -2.24 exceeds the critical value of ±1.96 for a 0.05 significance level, thus the null hypothesis is rejected. The company's claim is not supported by the sample data.

Here's the interpretation of the results:

1. **Z-Score:**
   - The Z-Score of -2.24 indicates that the sample mean of 4.5 purchases per customer is 2.24 standard deviations below the claimed population mean of 5 purchases per customer.

2. **P-Value:**
   - A P-Value less than 0.05 suggests that the observed data is statistically significant and not likely due to chance, assuming the null hypothesis is true.

3. **Decision:**
   - Since the P-Value is less than the significance level of 0.05, we reject the null hypothesis.

4. **Interpretation:**
   - Rejecting the null hypothesis suggests that there is significant evidence to conclude that the average number of purchases per customer is not 5 as claimed by the company.
   - In other words, the data provides evidence to suggest that the loyalty program has not increased the average monthly purchases per customer to 5.
   - The negative Z-Score indicates that the sample mean is below the claimed population mean.

5. **Implication:**
   - The company may need to re-evaluate the effectiveness of the loyalty program or consider other strategies to increase customer purchases.

6. **Confidence Level:**
   - With a significance level of 0.05, we can say with 95% confidence (100% - 5%) that the true population mean of monthly purchases per customer is not 5.

This interpretation provides a statistical insight into the effectiveness of the loyalty program based on the sample data, aiding in data-driven decision-making for the company.

**Question what is a 0.05 significance level?**

The 0.05 significance level acts as a cut-off for determining whether the observed difference in average purchases per customer is statistically significant or just a product of random variation. By setting a significance level of 0.05, the company accepts a 5% risk of concluding that the loyalty program had an effect when it actually did not (a false positive). Conversely, a 0.05 significance level implies a 95% confidence level, meaning the company can be 95% confident that the decision to reject or fail to reject the null hypothesis is correct.

**Question what is  p-value?**

The p-value provides a measure of statistical significance, helping the company understand the effectiveness of the loyalty program in increasing customer purchases.A p-value of 0.03 suggests that there's only a 3% chance of observing the sample data (or more extreme data) when the null hypothesis is true. This low probability indicates that the observed data is unlikely to have occurred by random chance alone, and there's a significant difference in the average monthly purchases per customer from the claimed value of 5. The p-value also helps in understanding the probability of a Type I error (false positive) – concluding that the loyalty program had an effect when it did not. In this case, a p-value of 0.03 means there's a 3% chance of making a Type I error if the null hypothesis is rejected.

```python
from scipy import stats
import math

# Given data
sample_mean = 4.5  # average purchases per customer from sample
population_mean = 5  # claimed average purchases per customer
sample_std_dev = 0.5  # standard deviation of purchases per customer from sample
sample_size = 50  # number of customers in the sample

# Calculating the z-score
z_score = (sample_mean - population_mean) / (sample_std_dev / math.sqrt(sample_size))

# Getting the p-value for two-tailed test
p_value = stats.norm.sf(abs(z_score)) * 2  # multiply by 2 for two-tailed test

print(f" p_value isL {round(p_value,11)}")
```

**Question what is Confidence Level?**

The confidence level represents the probability that a given interval estimate will contain the true population parameter. It's complementary to the significance level. For a significance level of 0.05 (5%), the corresponding confidence level is 95% (100% - 5%). This means that if you were to take many samples and build a confidence interval from each of them, about 95% of those intervals would contain the true population parameter (in this case, the true average monthly purchases per customer). Based on the Z-test result, and using a 0.05 significance level, we rejected the null hypothesis that the loyalty program increased the average monthly purchases per customer to 5. Given this decision:
- We can say with 95% confidence that the true average monthly purchases per customer, after implementing the loyalty program, is not 5.
- This doesn't mean there's a 95% probability the true mean is within our sample's interval; rather, if we were to sample many times, we expect about 95% of those sample intervals to contain the true mean.

The 95% confidence level gives the company a degree of certainty about their decision based on the sample data. It tells the company that the likelihood of their decision being correct (about the loyalty program's effect on purchases) is very high, and there's only a 5% chance that their decision to reject the null hypothesis is due to random sampling variability.

**Type I error** occurs when the null hypothesis is rejected even though it is true. In other words, it is a false positive result. This type of error is also known as a false alarm or a alpha error. It is denoted by the symbol alpha (α).

Imagine the company analyzing the effect of their loyalty program on average monthly purchases. This means that the analysis incorrectly rejected the null hypothesis that the loyalty program did not increase the average monthly purchases to 5, even though it actually did not.

Imagine the company analyzing the effect of their loyalty program on average monthly purchases. **Question: What if we are not interested in population but we are interested in comparing 2 groups. In other words A variant vs B variant? What can we do?**

### AB testing

In an A/B test, you create two versions of an element and randomly show them to users. You then monitor user interactions with each version and compare the results to determine which version is more effective.

**Let's generate random data test on it and then try on real one.**

Imagine you are a retail store manager and you're testing two different layouts to see which one encourages customers to spend more money. You have two versions: version A is the current layout, and version B is a new layout with more prominent product displays and better signage.

You collect data over a period of time on the average amount of money spent per customer visit in two sets of stores (or in the same store with different layouts over time). The data is normally distributed with a mean and standard deviation.

For Version A (current layout): The average amount spent per customer visit is $10, with a standard deviation of $2.
For Version B (new layout): The average amount spent per customer visit is $12, with a standard deviation of $2.

```python
import numpy as np

# Generating the data
version_A = np.random.normal(loc=10, scale=2, size=1000) # let's say this one is what we have.
version_B = np.random.normal(loc=12, scale=2, size=1000) # has higher mean about 20%.
```

**Performing the test using the T test.**

```python
from scipy.stats import ttest_ind

# Performing the t-test
t, p = ttest_ind(version_A, version_B)

# Printing the result
print(f"t = {t:.3f}")
print(f"p = {p:.3f}")
```

 - The p-value of 0.000 indicates that the observed differences between the two versions are statistically significant and that the difference between the means is very unlikely to be due to chance. This result suggests that there is a statistically significant difference between the two versions being tested.

- The negative value of t indicates that the mean of version A is smaller than the mean of version B. However, the absolute value of t is very large, indicating that there is a large difference between the means of the two samples.

- In summary, we can conclude that version B is statistically superior to version A based on the results of the A/B test. However, it is important to remember that the result of the A/B test is just a tool to aid decision-making and that other considerations, such as context and target audience, should also be taken into account.

Version B: A new, more vibrant design with animated elements.
For the experiment, each ad is shown to 500 different users (a total of 1000 users).

```python
from scipy.stats import norm
import numpy as np
# Let's define our sample size
n=500
p_a = .005  # real underlying probability of each trial for version A
p_b = .0075 # same for B
# We generate the random binomial samples (number of clicks for 500 ad impressions in each group)
s_a = np.random.binomial(n, p_a, 1)
s_b = np.random.binomial(n, p_b, 1)
# calculate pooled p
p=(sum(s_b)+sum(s_a))/(2*n)
print(p)
# calculate standard error
se=np.sqrt(p*(1-p)*(2/n))
print(se)
# calculate test statistic
z=(s_b/n-s_a/n)/se
print(z)
# get the p-value
print(1-norm.cdf(z))
```

**We cannot reject the null hypothesis, even though we know we should, as the samples were generated with known underlying probabilities (and pB was higher than pA).**

**Question: What we can do?**

Increase the sample size to increase the chance of not being wrong or in other words overcoming Type 2 error.

```python
# Define a function to calculate p-value based on sample size
def calculate_p_value(sample_size):
    s_a = np.random.binomial(sample_size, p_a, 1)
    s_b = np.random.binomial(sample_size, p_b, 1)
    p = (sum(s_b) + sum(s_a)) / (2 * sample_size)
    se = np.sqrt(p * (1 - p) * (2 / sample_size))
    z = (s_b / sample_size - s_a / sample_size) / se
    return 1 - norm.cdf(z)

# Range of sample sizes
sample_sizes = list(range(500, 20001, 100))  # from 500 to 10,000 in steps of 500

# Calculate p-values for each sample size
p_values = [calculate_p_value(n) for n in sample_sizes]

# Plotting the results
plt.figure(figsize=(10,6))
plt.plot(sample_sizes, p_values, marker='o', linestyle='-')
plt.axhline(y=0.05, color='r', linestyle='--', label='Significance Threshold (0.05)')
plt.xlabel('Sample Size')
plt.ylabel('P-value')
plt.title('P-value vs. Sample Size')
plt.legend()
plt.grid(True)
plt.show()
```

Based on this graph we can say after the 12000 sample size we will be able to make a decision and probability of wrong decision is low.

**Question: But we actually need to define the sample size before the test. How are doing it?**

The formula to calculate a big enough sample relies on the concept of effect size, or: what is the minimum difference you want to be able to detect (and that makes business sense)? It could be, for instance, a difference of at least 0.05 percentage point, bringing an existing conversion rate of 0.1% up to 0.15%. Of course, the required effect size would vary depending on the business sector and the applications, but it needs to be somewhat justifiable to the management teams.

The formula to calculate the sample size uses the significance level (alpha), the power level (1-beta) and the effect size (ES), you can find a very good explanation on how the formula is derived here (https://online.stat.psu.edu/stat415/lesson/25/25.3), we are essentially looking for a value c that is both a cut-off value for a rejection region and for our probability of getting a false negative (see figure below)

The formula for calculating the required sample size for a hypothesis test comparing two proportions (like conversion rates) can be derived from the formula for the z-score. The formula for the required sample size \( n \) for each group is:

$$ n = \left( \frac{Z_{\alpha/2} + Z_{\beta}}{\sqrt{2p(1-p)} \times (p_1 - p_2)} \right)^2 $$

Where:
- $ Z_{\alpha/2} $ is the critical value from the Z-distribution for a given significance level $ \alpha $. For a two-sided test with $ \alpha = 0.05 $, $ Z_{\alpha/2} $ is approximately 1.96.
- $ Z_{\beta} $ is the Z-value that corresponds to the desired power of the test. For example, for a power of 0.80, $ Z_{\beta} $ is approximately 0.84.
- $ p $ is the pooled proportion, which is calculated as $ p = \frac{p_1 + p_2}{2} $.
- $ p_1 $ and $ p_2 $ are the conversion rates (or proportions) of the two groups you're comparing.

This formula gives the sample size required for each group to detect a difference of $ p_1 - p_2 $ between the two proportions with a given level of power and significance.

It should be easier with an example: you are running a banner (A) campaign with a CTR of 0.5% and you want to test a new banner (B). You are only interested in a minimum 50% increase in CTR, so a new improved CTR of 0.5%*1.5=0.75%, an absolute difference of 0.25 percentage points. What is the minimum number of impressions you need to run in each version to be at least 80% confident to detect a difference at the 5% significance level?

a) The significance level is 5% so our Z(1-alpha) value is 1.645
b) The required Power is 80% so our Z(1-beta) is 0.842
c) Our overall (average) proportion p is (0.005+0.0075)/2=0.00625
d) And our effect size ES is (0.0075–0.005)/sqrt(0.00625*0.99375) =0.03172

So our n = 2*[(1.645+0.842)/0.03172]² = 12,296 impressions on each banner

In practice you can use either calculators as this
https://www.evanmiller.org/ab-testing/sample-size.html

Or the rull of thumb version to calculate n for a power of 80% and a significance level of 5% with delta as minimum difference.

$$n = 16 * \frac{p(1-p)}{\delta^2}$$

### Bayesian A/B Testing: A Detailed Introduction

**Basics of Bayesian Inference**

Bayesian inference is a method of statistical inference where we use probability to represent uncertainty about our model parameters. The foundation of Bayesian statistics is Bayes' theorem:

$$ P(A|B) = \frac{P(B|A) \times P(A)}{P(B)} $$

Where:
- $ P(A|B) $ is the posterior probability of A given B.
- $ P(B|A) $ is the likelihood.
- $ P(A) $ is the prior probability of A.
- $ P(B) $ is the evidence.

In the context of A/B testing:

- **Prior, $ P(A) $**: Our beliefs about the conversion rate before seeing new data.
  
- **Likelihood, $ P(B|A) $**: The probability of observing the data given our model.
  
- **Posterior, $ P(A|B) $**: Updated belief about the conversion rate after seeing the data.

## Numeric example

### Bayesian Inference in Medical Testing:

**Context**:
You're a doctor considering the likelihood that a patient has a rare disease. Only 1% of the population has this disease. You administer a test that, in the past, has been 99% accurate for those with the disease (sensitivity) and 95% accurate for those without the disease (specificity).

**Scenario**:
Your patient tests positive. What's the probability they actually have the disease?

**Bayesian Setup**:
1. **Prior Probability (P(Disease))**: 1% or 0.01. This is our initial belief about the patient having the disease before testing.
2. **Likelihood (P(Positive | Disease))**: 99% or 0.99. This is the probability of a positive test given that someone has the disease.
3. **Evidence (P(Positive))**: This is the total probability of a positive test. It can be calculated as:
   $$ P(Positive) = P(Disease) \times P(Positive | Disease) + P(No Disease) \times P(Positive | No Disease) $$
   $$ = 0.01 \times 0.99 + 0.99 \times 0.05 = 0.0594 $$

**Application of Bayes' Theorem**:
$$ P(Disease | Positive) = \frac{P(Positive | Disease) \times P(Disease)}{P(Positive)} $$
$$ = \frac{0.99 \times 0.01}{0.0594} $$
$$ \approx 0.166 $$

**Result**:
Given the positive test result, there's approximately a 16.6% chance the patient actually has the disease, even though the test itself is quite accurate. This seemingly low probability despite a positive test result arises because the disease is rare in the population.

This example illustrates the power of Bayesian inference in considering both the prior context and the new evidence to make informed decisions.


**Application to A/B Testing**

Let's say we want to test two versions of a website: A and B.

1. **Set Up Priors**:
   - Assume a Beta distribution for both A and B based on historical data or beliefs.
   - $ A \sim Beta(\alpha_A, \beta_A) $
   - $ B \sim Beta(\alpha_B, \beta_B) $
   
   The Beta distribution is commonly used in Bayesian A/B testing because it is conjugate to the binomial distribution, which makes the math easier.

2. **Gather Data**:
   - Let $ x_A $ be the number of successes (e.g., conversions) for variant A and $ n_A $ be the total number of trials for A.
   - Similarly, let $ x_B $ and $ n_B $ be the number of successes and trials for variant B.

3. **Calculate Posterior Distributions**:
   - The posterior distribution for A after observing the new data is:
     $$ A' \sim Beta(\alpha_A + x_A, \beta_A + n_A - x_A) $$
   - Similarly, for B:
     $$ B' \sim Beta(\alpha_B + x_B, \beta_B + n_B - x_B) $$

4. **Decision Making**:
   - Compute the probability that A is better than B by simulating a large number of draws from the posterior distributions and comparing them.
   - For example:
     $ P(A' > B') $
   This gives the probability that variant A has a higher conversion rate than variant B.

**Advantages of Bayesian A/B Testing**:

- **Direct Probabilities**: Instead of p-values, you get probabilities which are easier to interpret.
  
- **Flexibility**: No need for fixed sample sizes. You can stop the test as soon as you're confident in the results.

- **Incorporate External Information**: Bayesian methods allow for the incorporation of prior knowledge.

**Conclusion**:

Bayesian A/B testing uses prior beliefs, updates them with data, and provides direct probabilities for decision-making. This approach is flexible, interpretable, and can often lead to faster conclusions than traditional testing.

**Now that we understand how to select a distribution for our experiment, let’s learn how to determine experimental impact. We do this by calculating three statistics:**

1. Treatment lift
2. Probability of being the best
3. Expected loss

```python
# Defining the sample sizes and conversion counts for both control and treatment groups
control_sample = 500  # Total number of samples in the control group
control_converted = 100  # Number of successful conversions in the control group
treatment_sample = 500  # Total number of samples in the treatment group
treatment_converted = 120  # Number of successful conversions in the treatment group
```

### Treatment Lift

For our first calculation, treatment lift is simply the percent change between our treatment and control. This is our treatment impact and can be calculated with the formula below. And, for simplicity, let’s assume there’s just one treatment and one control.
$$ treatmentlift = (treatment - control) / control $$

### Probability of being the best

For our second and third calculations, we need to develop a simulation aka Monte Carlo simulation. Why you might ask? Well, because we only have one sample of data, we can’t compute any probabilities — we don’t know how the data would look in alternative samples. So, to generate more data, we leverage knowledge about our data generating mechanism (the posterior distribution) and samples from that. 10,000 samples is a good rule of thumb.


For the probability of being best, our second statistic, we simply look at all of our simulated samples and calculate the percentage of the time that our treatment is better than our control. That proportion becomes our probability of being best for the treatment. Check out the pythonic pseudo code below.

```python
# Importing necessary libraries
import numpy as np
from scipy.stats import beta
import matplotlib.pyplot as plt



# Setting up the parameters for the prior distribution.
prior_alpha = 1  # Alpha parameter for the prior Beta distribution
prior_beta = 1  # Beta parameter for the prior Beta distribution

# Defining the prior Beta distribution
prior = beta(prior_alpha, prior_beta)  # Prior Beta distribution

# Updating the prior based on the observed data for the control group
# Adding the number of successful conversions to alpha and the number of failures to beta
control = beta(prior_alpha + control_converted, prior_beta + control_sample - control_converted)

# Updating the prior based on the observed data for the treatment group
# Adding the number of successful conversions to alpha and the number of failures to beta
treatment = beta(prior_alpha + treatment_sample, prior_beta + treatment_converted - treatment_sample)
```

```python
# Importing the decimal library to set precision for floating point arithmetic
import decimal
# Setting the precision for decimal floating point arithmetic to 2 digits
decimal.getcontext().prec = 2

# Importing the numpy library
import numpy as np

# Simulating 10,000 samples from the posterior Beta distribution for the control group
control_simulation = np.random.beta(
    prior_alpha + control_converted,
    prior_beta + control_sample - control_converted,
    size=10000
)

# Simulating 10,000 samples from the posterior Beta distribution for the treatment group
treatment_simulation = np.random.beta(
    prior_alpha + treatment_converted,
    prior_beta + treatment_sample - treatment_converted,
    size=10000
)

# Comparing the simulated samples to calculate the proportion of times
# the treatment group beats the control group
treatment_won = [i <= j for i, j in zip(control_simulation, treatment_simulation)]

# Calculating the mean of the comparison to get the probability of the treatment
# group beating the control group
chance_of_beating_control = np.mean(treatment_won)

# Printing the chance of the treatment group beating the control group
# with high precision using the decimal library
print(f'Chance of treatment beating control is {decimal.getcontext().create_decimal(chance_of_beating_control)}')
```

Let's visualize both

### Loss threshhold

**Question: How are we gonna calculate the loss from the data?**

### Expected loss

Simple approach will be to calculate how much we lost in the cases when we chose treatment and vise versa given the simulations we have. There are more complex approaches like defining loss function but we will not use them for now.

```python
# Set the precision for decimal calculations to 10 decimal places.
decimal.getcontext().prec = 10

# Calculate the loss for the control scenario for each simulation pair.
# The loss is defined as the maximum of (treatment value - control value, 0).
loss_control = [np.max(j - i, 0) for i,j in zip(control_simulation, treatment_simulation)]

# Calculate the loss for the treatment scenario for each simulation pair.
# The loss is defined as the maximum of (control value - treatment value, 0).
loss_treatment = [np.max(i - j, 0) for i,j in zip(control_simulation, treatment_simulation)]

# Aggregate the loss for the control scenario based on whether the treatment was won or not.
# If the treatment was won (i is 1), the loss for the control scenario is counted, otherwise, it's 0.
all_loss_control = [int(i)*j for i,j in zip(treatment_won, loss_control)]

# Aggregate the loss for the treatment scenario based on whether the treatment was won or not.
# If the treatment was not won (i is 0), the loss for the treatment scenario is counted, otherwise, it's 0.
all_loss_treatment = [(1 - int(i))*j for i,j in zip(treatment_won, loss_treatment)]

# Calculate the expected loss for the control scenario by taking the mean of all aggregated losses.
expected_loss_control = np.mean(all_loss_control)

# Calculate the expected loss for the treatment scenario by taking the mean of all aggregated losses.
expected_loss_treatment = np.mean(all_loss_treatment)

# Print the expected loss for both the control and treatment scenarios with a precision of 10 decimal places.
print(f'Expected loss of choosing control: {decimal.getcontext().create_decimal(expected_loss_control)}. Expected loss of choosing treatment: {decimal.getcontext().create_decimal(expected_loss_treatment)}')
```

Based on this we would say Chance of treatment beating control is 94%. considereing the expected loss of choosing treatment is 0.0007 which is lower than our loss threshold 0.001. We can go with second variant.

Bayesian A/B testing is a statistical method used to compare different versions of a product, like two different web pages, to find out which one performs better with respect to some metric. Unlike traditional frequentist A/B testing, Bayesian A/B testing incorporates prior beliefs and provides a probabilistic framework for decision making. Here are the steps involved in Bayesian A/B testing:

1. **Formulate the Problem:**
   - Define the metric you want to improve (e.g., click-through rate, conversion rate, etc.)
   - Determine the variations you want to test (e.g., version A and version B of a webpage).

2. **Set Up Prior Distributions:**
   - Establish prior beliefs about the parameters you are estimating (e.g., the conversion rates for version A and B).
   - Common choices for priors include beta distributions, which are conjugate priors for binomial likelihoods.

3. **Collect Data:**
   - Run the experiment and collect data on how users interact with the different versions of your product.

4. **Calculate Likelihood:**
   - Based on the data, calculate the likelihood of observing the data under different parameter values.

5. **Compute Posterior Distributions:**
   - Update your prior beliefs with the likelihood to compute the posterior distributions of the parameters using Bayes' theorem.

6. **Calculate Expected Loss:**
   - Define a loss function to quantify the cost of making a wrong decision (e.g., the cost of choosing the worse performing version).
   - Calculate the expected loss for each decision (e.g., choosing version A or B) by integrating the loss function over the posterior distributions.

7. **Decision Making:**
   - Make your decision based on the expected loss calculations. Typically, you would choose the version that minimizes the expected loss.

8. **Interpretation and Reporting:**
   - Interpret the results of the Bayesian A/B test in the context of your problem.
   - Report the results to stakeholders, including the posterior distributions, expected loss, and the decision made.

9. **Iterate:**
   - If necessary, iterate on your A/B test by incorporating feedback, adjusting your variations, or collecting more data.

10. **Implementation:**
    - Implement the winning version, monitor the performance, and ensure the results align with the findings from the Bayesian A/B testing.
