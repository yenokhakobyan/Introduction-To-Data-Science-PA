## Sampling Bias

In statistics, sampling bias is a bias in which a sample is collected in such a way that some members of the intended population are less likely to be included than others. It results in a biased sample, a non-random sample of a population (or non-human factors) in which all individuals, or instances, were not equally likely to have been selected. If this is not accounted for, results can be erroneously attributed to the phenomenon under study rather than to the method of sampling.



Sampling bias occurs when a sample is not representative of the population from which it was drawn, leading to biased and potentially inaccurate conclusions.

1. **Self-selection Bias**:
    - **Example**: Suppose a university sends out a satisfaction survey via email to all of its students to gauge overall student satisfaction with the university experience. However, only students who feel strongly about their experiences (either positively or negatively) take the time to respond to the survey. This self-selection leads to a biased sample that doesn't accurately represent the views of the entire student body. The results might overestimate both the satisfaction and dissatisfaction levels among all students.

2. **Convenience Sampling Bias**:
    - **Example**: A researcher wants to estimate the average household income in a city. They decide to conduct surveys at a local high-end shopping mall, where it's convenient for them to approach people. However, the individuals frequenting this mall likely have higher incomes compared to the general population of the city. As a result, the sample obtained is biased, and the estimate of the average household income derived from this sample is likely to be higher than the actual average income in the city.

3. **Non-response Bias**:
    - **Example**: In a health study, researchers send out surveys to a random sample of individuals to collect data on various health behaviors. However, individuals with certain health conditions might be less likely to respond due to the sensitive nature of the questions or their health status. The lack of responses from these individuals can lead to non-response bias, making the sample unrepresentative of the overall population and potentially skewing the results of the study.

4. **Undercoverage Bias**:
    - **Example**: In a political poll aiming to predict election outcomes, a polling agency uses landline phones to contact potential respondents. However, this method of contact may exclude younger voters who are more likely to use mobile phones exclusively. This undercoverage of younger voters can lead to a biased sample that doesn’t accurately reflect the population’s voting intentions, potentially leading to incorrect predictions regarding the election outcome.

### Types Of Sampling Methods

- Simple Random Sampling (SRS)

- Stratified Sampling

- Cluster Sampling

- Systematic Sampling

**Simple Random Sampling**:

In a simple random sample (SRS) of a given size, all such subsets of the frame are given an equal probability. Each element of the frame thus has an equal probability of selection: the frame is not subdivided partitioned. Any given pair of elements has the same chance of selection as any other such pair (and similarly for triples, and so on). This minimizes bias and simplifies analysis of results.In particular, the variance between individual results within the sample is a good indicator of variance in the overall population, which makes it relatively easy to estimate the accuracy of results.

SRS can be vulnerable to sampling error because the randomness of the selection may result in a sample that doesn't reflect the makeup of the population.

**Stratified Sampling**:

It is possible when it makes sense to partition the population into groups based on a factor that may influence the variable that is being measured.   These groups are then called strata.  An individual group is called a stratum.  With stratified sampling one should:

- partition the population into groups (strata)

- obtain a simple random sample from each group (stratum)

- collect data on each sampling unit that was randomly sampled from each group (stratum)

**Cluster Sampling**:

It is very different from Stratified Sampling. With cluster sampling one should

- divide the population into groups (clusters).

- obtain a simple random sample of so many clusters from all possible clusters.

- obtain data on every sampling unit in each of the randomly selected clusters.

**Systematic Sampling**:

Systematic sampling (also known as interval sampling) relies on arranging the study population according to some ordering scheme and then selecting elements at regular intervals through that ordered list.Systematic sampling involves a random start and then proceeds with the selection of every kth element from then onwards.It is easy to implement and the stratification induced can make it. However, systematic sampling is especially vulnerable to periodicities in the list. If periodicity is present and the period is a multiple or factor of the interval used, the sample is especially likely to be unrepresentative of the overall population, making the scheme less accurate than simple random sampling.

Systematic sampling is that even in scenarios where it is more accurate than SRS, its theoretical properties make it difficult to quantify that accuracy.

## Quantitative and Qualitative data

Quantitative and qualitative data provide different outcomes, and are often used together to get a full picture of a population. For example, if data are collected on annual income (quantitative), occupation data (qualitative) could also be gathered to get more detail on the average annual income for each type of occupation.

Quantitative and qualitative data can be gathered from the same data unit depending on whether the variable of interest is numerical or categorical.

**Quantitative data**

- Information that can be handled numerically.  Quantitative variables are numerical variables: counts, percents, or numbers.

- Quantitative data always are associated with a scale measure.

- Probably the most common scale type is the ratio-scale. Observations of this type are on a scale that has a meaningful zero value but also have an equidistant measure

- Statistics that describe or summarise can be produced for quantitative data and to a lesser extent for qualitative data.

- As quantitative data are always numeric they can be ordered, added together, and the frequency of an observation can be counted. Therefore, all descriptive statistics can be calculated using quantitative data.

- By making inferences about quantitative data from a sample, estimate or projection for the total population can be produced.


- Quantitative data can be used to inform broader understandings of a population

**Qualitative data**

- Information that refers to the quality of something. Ethnographic research, participant observation, open-ended interviews, etc., may collect qualitative data. However, often there is some element of the results obtained via qualitative research that can be handled numerically, e.g., how many observations, number of interviews conducted, etc.

- Qualitative data are measures of 'types' and may be represented by a name, symbol, or a number code.

- Qualitative data are data about categorical variables (e.g. what type).

- Statistics that describe or summarise can be produced for quantitative data and to a lesser extent for qualitative data.

- As qualitative data represent individual (mutually exclusive) categories, the descriptive statistics that can be calculated are limited, as many of these techniques require numeric values which can be logically ordered from lowest to highest and which express a count.

- Qualitative data are not compatible with inferential statistics as all techniques are based on numeric values.

- It is used to consider how a population may change or progress into the future.



## Example of a dataset: A mobile application where users buy products.

P.S: This is actually a sample of the data. As the population can be defined as all the transaction made by world people. Or purchases done for the lifetime etc. but we have only several of them.

````python
import pandas as pd
import numpy as np
import random
from faker import Faker

# Set seed for reproducibility
np.random.seed(0)
random.seed(0)

# Initialize a Faker generator
fake = Faker()

# Generate user data
user_ids = range(1, 5353)  # 5352 user_ids
countries = [fake.country_code() for _ in range(5352)]
occupations = [fake.job() for _ in range(5352)]
account_creation_date = [fake.date_between_dates(date_start=pd.to_datetime('2022-01-01'), date_end=pd.to_datetime('2022-12-31')) for _ in range(5352)]

# Generate transaction data
transaction_counts = np.random.poisson(lam=8, size=len(user_ids))  # Average 8 transactions per user
transactions = []
for user_id, country, occupation, acc_date, trans_count in zip(user_ids, countries, occupations, account_creation_date, transaction_counts):
    for _ in range(trans_count):
        transaction_date = fake.date_between_dates(date_start=acc_date, date_end=pd.to_datetime('2022-12-31'))
        main_event = fake.random.choice(['Product View', 'Product Added', 'Checkout Initiated', 'Payment Completed'])
        event_value = round(np.random.exponential(scale=100), 2)  # Random event value, e.g., price of the product
        transactions.append([user_id, transaction_date, country, occupation, main_event, event_value])

# Create DataFrame
columns = ['user_id', 'date', 'country_code', 'occupation', 'main_event', 'event_value']
transaction_df = pd.DataFrame(transactions, columns=columns)

# Adding two more qualitative features: 'device_type' and 'payment_method'
transaction_df['device_type'] = np.random.choice(['Mobile', 'Tablet', 'Desktop'], size=len(transaction_df))
transaction_df['payment_method'] = np.random.choice(['Credit Card', 'Debit Card', 'PayPal', 'Apple Pay', 'Google Pay'], size=len(transaction_df))

transaction_df.loc[transaction_df['main_event'] != 'Payment Completed', 'event_value'] = 1

# Display the first few rows of the dataset
transaction_df.head()
````

## Descriptive Statistics

**Descriptive Statistics**

- Descriptive statistics give information that describes the data in some manner. For example, suppose a pet shop sells cats, dogs, birds and fish. If 100 pets are sold, and 40 out of the 100 were dogs, then one description of the data on the pets sold would be that 40% were dogs.

- This same pet shop may conduct a study on the number of fish sold each day for one month and determine that an average of 10 fish were sold each day. The average is an example of descriptive statistics.

- Some other measurements in descriptive statistics answer questions such as 'How widely dispersed is this data?', 'Are there a lot of different values?' or 'Are many of the values the same?', 'What value is in the middle of this data?', 'Where does a particular data value stand with respect with the other values in the data set?'

- A graphical representation of data is another method of descriptive statistics. Examples of this visual representation are histograms, bar graphs and pie graphs, to name a few. Using these methods, the data is described by compiling it into a graph, table or other visual representation.

- This provides a quick method to make comparisons between different data sets and to spot the smallest and largest values and trends or changes over a period of time. If the pet shop owner wanted to know what type of pet was purchased most in the summer, a graph might be a good medium to compare the number of each type of pet sold and the months of the year.



**Inferential statistics**

- Inferential statistics is one of the two main branches of statistics. Inferential statistics use a random sample of data taken from a population to describe and make inferences about the population.

- For instance, we use inferential statistics to try to infer from the sample data what the population might think. Or, we use inferential statistics to make judgments of the probability that an observed difference between groups is a dependable one or one that might have happened by chance in this study. Thus, we use inferential statistics to make inferences from our data to more general conditions; we use descriptive statistics simply to describe what's going on in our data.

- Properties of samples, such as the mean or standard deviation, are not called parameters, but statistics. Inferential statistics are techniques that allow us to use these samples to make generalizations about the populations from which the samples were drawn.

- It is, therefore, important that the sample accurately represents the population. The process of achieving this is called sampling (sampling strategies are discussed in detail here on our sister site). Inferential statistics arise out of the fact that sampling naturally incurs sampling error and thus a sample is not expected to perfectly represent the population.

- The methods of inferential statistics are (1) the estimation of parameter(s) and (2) testing of statistical hypotheses.

    Example 4: Hypothesis Testing - Are Transaction Values Significantly Different between Device Types?


````python
# ANOVA Test to compare means across device types
import statsmodels.api as sm
from statsmodels.formula.api import ols

model = ols('event_value ~ C(device_type)', data=transaction_df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
````

Conclusion:
The ANOVA test can tell us whether there's a statistically significant difference in the average transaction value between different device types. A low p-value (e.g., <0.05) would indicate a significant difference, suggesting that the device type might affect the transaction value.

In probability and statistics, a random variable, random quantity, aleatory variable, or stochastic variable is a variable whose possible values are outcomes of a random phenomenon. ### Distribution:
The distribution function \( F(x) \) of a discrete random variable \( X \) gives the probability that \( X \) is less than or equal to \( x \), for every value \( x \).

Let's assume the following probabilities for each value of \( X \):
- \( P(X = 1) = 0.1 \)
- \( P(X = 2) = 0.2 \)
- \( P(X = 3) = 0.3 \)
- \( P(X = 4) = 0.4 \)

Now, the distribution function \( F(x) \) is found by summing up the probabilities:
- \( F(1) = P(X \leq 1) = 0.1 \)
- \( F(2) = P(X \leq 2) = 0.1 + 0.2 = 0.3 \)
- \( F(3) = P(X \leq 3) = 0.1 + 0.2 + 0.3 = 0.6 \)
- \( F(4) = P(X \leq 4) = 0.1 + 0.2 + 0.3 + 0.4 = 1.0 \)

### Mean:
The mean \( \mu \) of a discrete random variable \( X \) is calculated as a weighted average of the possible values that \( X \) can take, where each value is weighted by its probability:
$$ \mu = \sum [x_i \cdot p(x_i)] = (1 \cdot 0.1) + (2 \cdot 0.2) + (3 \cdot 0.3) + (4 \cdot 0.4) = 3.0 $$

### Standard Deviation:
The standard deviation \( \sigma \) of a discrete random variable \( X \) measures the spread of the distribution, and is calculated as:
$$ \sigma = \sqrt{\sum [(x_i - \mu)^2 \cdot p(x_i)]} = \sqrt{(1 - 3)^2 \cdot 0.1 + (2 - 3)^2 \cdot 0.2 + (3 - 3)^2 \cdot 0.3 + (4 - 3)^2 \cdot 0.4} $$

Now, let's compute the mean and standard deviation in python:

````python
# Given probabilities and values
x_values = np.array([1, 2, 3, 4])
probabilities = np.array([0.1, 0.2, 0.3, 0.4])

# Calculating the mean
mean = np.sum(x_values * probabilities)

# Calculating the variance
variance = np.sum(((x_values - mean)**2) * probabilities)

# Calculating the standard deviation
std_dev = np.sqrt(variance)

(mean, std_dev)
````

## 1. Bernoulli Distribution

### Definition:

A Bernoulli distribution is a discrete probability distribution for a Bernoulli trial — a random experiment that has exactly two possible outcomes: success (often labeled as 1) and failure (often labeled as 0). The single parameter of a Bernoulli distribution is \( p \), the probability of success.

The probability mass function (PMF) of a Bernoulli distribution is given by:
$$ P(X = x) = p^x(1 - p)^{1-x} \quad \text{for} \quad x \in \{0, 1\}. $$


where:
- \( X \) is the random variable representing the outcome (1 for success and 0 for failure),
- \( p \) is the probability of success.

### Example:

Suppose we are flipping a biased coin where the probability of landing on heads (success) is 0.7 and the probability of landing on tails (failure) is 0.3. This scenario can be modeled by a Bernoulli distribution with parameter \( p = 0.7 \).

Let's compute the probability of getting a head (success) and a tail (failure) using the Bernoulli distribution formula:

````python
import sympy as sp  # Importing the symbolic computation library to handle the formula

# Given parameter
p = 0.7
x_values = sp.symbols('x')  # Defining the variable x as a symbol

# Bernoulli distribution formula
bernoulli_formula = p**x_values * (1 - p)**(1 - x_values)

# Computing the probabilities
probability_success = bernoulli_formula.subs(x_values, 1)  # Substituting x = 1 for success
probability_failure = bernoulli_formula.subs(x_values, 0)  # Substituting x = 0 for failure

probability_success, probability_failure
````

#### Mean:

The formula for the mean (or expected value) of a Bernoulli distribution is straightforward and is given by:

$$ \mu = p $$

where:
- \( \\mu \) is the mean of the Bernoulli distribution,
- \( p \) is the probability of success on a single trial.

In a Bernoulli distribution, each trial only has two possible outcomes: success (often coded as \(1\)) or failure (often coded as \(0\)). The mean of the distribution is simply the probability of success \( p \).

#### Variance:
The variance (\( \sigma^2 \)) of a Bernoulli distribution is given by the formula:

$$ \sigma^2 = E[(X - \mu)^2] $$

where \( E \) denotes the expected value, and \( \mu \) is the mean of the random variable \( X \).

For a Bernoulli distribution, the random variable \( X \) can take on two values: 1 (success) with probability \( p \), and 0 (failure) with probability \( 1 - p \). The mean \( \mu \) of a Bernoulli distribution is \( p \).
$
Now, we'll substitute these values into the formula for variance:

$$ \sigma^2 = E[(X - p)^2] $$

$$ = (1 - p)^2 \cdot P(X = 0) + (1 - p)^2 \cdot P(X = 1) $$

$$ = (1 - p)^2 \cdot (1 - p) + (1 - p)^2 \cdot p $$

$$ = (1 - p)^3 + (1 - p)^2 \cdot p $$

$$ = (1 - p) \cdot [(1 - p)^2 + (1 - p) \cdot p] $$

$$ = (1 - p) \cdot [(1 - p) \cdot (1 - p + p)] $$

$$ = (1 - p) \cdot [(1 - p) \cdot 1] $$

$$ = (1 - p) \cdot (1 - p) $$

$$ = p(1 - p) $$

#### Standard Deviation:
The standard deviation \( \sigma \) is the square root of the variance \( \sigma^2 \), so we have:

$$ \sigma = \sqrt{p(1 - p)} $$


## 2. Binomial Distribution

A binomial distribution is a discrete probability distribution of the number of successes in a sequence of \( n \) independent experiments, each asking a yes–no question. Each experiment (also known as Bernoulli trial) results in a success with probability \( p \) or a failure with probability \( 1-p \). The number of successes is denoted by \( X \), and the distribution of \( X \) is called a binomial distribution.

#### Probability Mass Function (PMF):
The probability mass function of a binomial distribution is given by:

$$ P(X = k) = \binom{n}{k} \cdot p^k \cdot (1-p)^{n-k} $$

where:
- \( n \) is the number of trials,
- \( k \) is the number of successes,
- \( p \) is the probability of success on a single trial,
- \( \binom{n}{k} \) is the binomial coefficient, representing the number of ways to choose \( k \) successes out of \( n \) trials.

#### Mean and Variance:
The mean (\( \mu \)) and variance (\( \sigma^2 \)) of a binomial distribution are given by:

$$\mu = n \cdot p $$
$$ \sigma^2 = n \cdot p \cdot (1 - p) $$

### Connection between Bernoulli and Binomial Distributions:

A binomial distribution can be thought of as the sum of \( n \) independent and identically distributed (i.i.d.) Bernoulli random variables, each with probability of success \( p \).

In other words, if you have a Bernoulli distribution with probability \( p \) of success, and you conduct \( n \) independent trials, then the number of successes \( X \) follows a binomial distribution with parameters \( n \) and \( p \).

### Summary:
- A Bernoulli distribution is a special case of a binomial distribution where the number of trials \( n = 1 \).
- A binomial distribution extends the Bernoulli distribution to more than one trial.
- Each trial in a binomial distribution is a Bernoulli trial.

In essence, while a Bernoulli distribution represents a single trial with two possible outcomes, a binomial distribution represents a series of such trials and counts the number of successes.

The derivation of the mean and standard deviation for a binomial distribution follows from the properties of the distribution and the nature of a Bernoulli trial.

### Mean of Binomial Distribution:
The mean (expected value) of a binomial distribution is derived from the expectation of the sum of independent random variables.

Let \( X \) be the sum of \( n \) independent Bernoulli trials, each with probability \( p \) of success:

$$ X = X_1 + X_2 + \ldots + X_n $$

where each \( X_i \) is a Bernoulli random variable with \( P(X_i = 1) = p \) and \( P(X_i = 0) = 1 - p \).

The expected value of \( X \) is the sum of the expected values of each \( X_i \):

$$ E[X] = E[X_1 + X_2 + \ldots + X_n] = E[X_1] + E[X_2] + \ldots + E[X_n] = n \cdot p $$

So, the mean \( \mu \) of a binomial distribution is \( n \cdot p \).

### Variance of Binomial Distribution:
The variance of a binomial distribution is derived from the variance of the sum of independent random variables.

The variance of \( X \) is the sum of the variances of each \( X_i \):

$$ Var[X] = Var[X_1 + X_2 + \ldots + X_n] = Var[X_1] + Var[X_2] + \ldots + Var[X_n] = n \cdot p(1 - p) $$

So, the variance \( \sigma^2 \) of a binomial distribution is \( n \cdot p(1 - p) \).

### Standard Deviation of Binomial Distribution:
The standard deviation \( \sigma \) is the square root of the variance:

$$ \sigma = \sqrt{n \cdot p(1 - p)} $$

These formulas for the mean, variance, and standard deviation of a binomial distribution are derived based on the properties of the Bernoulli trials that constitute the binomial distribution, and the properties of expectation and variance for sums of independent random variables.

## 3. Normal distribution

Normal distribution represents the behavior of most of the situations in the universe (That is why it’s called a “normal” distribution. I guess!). The large sum of (small) random variables often turns out to be normally distributed, contributing to its widespread application. Any distribution is known as Normal distribution if it has the following characteristics:

- The mean, median and mode of the distribution coincide.

- The curve of the distribution is bell-shaped and symmetrical about the line x=μ.

- The total area under the curve is 1.

- Exactly half of the values are to the left of the center and the other half to the right.


The Normal distribution is a continuous probability distribution characterized by its mean (\( \mu \)) and variance (\( \sigma^2 \)). It's one of the most important and widely used distributions in statistics.

### Mean of Normal Distribution:
The mean (\( \mu \)) of a normal distribution is the central value around which the data is symmetrically distributed. It's where the peak of the distribution curve occurs. The formula for the mean of a normal distribution is straightforward:

$$ \mu = \frac{1}{N} \sum x_i $$

where \( N \) is the number of data points, and \( x_i \) are the individual data points.

### Variance and Standard Deviation of Normal Distribution:
The variance (\( \sigma^2 \)) of a normal distribution measures the dispersion of the data points from the mean. It's calculated as:

$$ \sigma^2 = \frac{1}{N} \sum (x_i - \mu)^2 $$

The standard deviation (\( \sigma \)) is simply the square root of the variance:

$$ \sigma = \sqrt{\frac{1}{N} \sum (x_i - \mu)^2} $$

## Connection to Binomial Distribution
When the number of trials in a binomial distribution is large, the distribution of outcomes begins to resemble a normal distribution. This is known as the Central Limit Theorem. The mean and variance of the binomial distribution (\( np \) and \( np(1-p) \) respectively) become the mean and variance of the normal distribution.

### Visualization:
A normal distribution is symmetrical and its graph is bell-shaped. The mean, median, and mode of a normal distribution are all equal and located at the center of the distribution.


````python

import numpy as np
import matplotlib.pyplot as plt

# Given parameters for a normal distribution
mu = 0  # Mean
sigma = 1  # Standard deviation

# Generating values for X between -5 and 5
x_values = np.linspace(-5, 5, 400)

# Calculating the PDF for each value of X
pdf_values = (1/(sigma * np.sqrt(2*np.pi))) * np.exp( - (x_values - mu)**2 / (2 * sigma**2) )

# Plotting the distribution
plt.plot(x_values, pdf_values)
plt.title('Normal Distribution')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.show()
````

## Connection between Binominal and Normal Distribution

The binomial distribution tends to the normal distribution as the number of trials (\( n \)) increases, given that \( np(1-p) \) is sufficiently large. This is an example of the Central Limit Theorem in action. The mean (\( \mu \)) and variance (\( \sigma^2 \)) of the binomial distribution become the mean and variance of the normal distribution, respectively.

$$ \mu = n \cdot p $$
$$ \sigma^2 = n \cdot p \cdot (1-p) $$

When \( n \) is large, a binomial distribution with parameters \( n \) and \( p \) can be approximated by a normal distribution with mean $ \mu = n \cdot p $ and variance $ \sigma^2 = n \cdot p \cdot (1-p) $.

Let's illustrate this with an example. Suppose we have a binomial distribution with \( p = 0.5 \) (fair coin flips), and we'll look at how the distribution changes as \( n \) increases.


In this code:
- We first define the probability of success \( p \) and an array of different numbers of trials \( n \).
- We then loop over the different values of \( n \) to plot the probability mass function (PMF) of the binomial distribution and the probability density function (PDF) of the normal distribution.
- As \( n \) increases, the binomial distribution becomes more similar to the normal distribution, illustrating the convergence.


````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom, norm

p = 0.5  # Probability of success

# Array of different number of trials
n_values = [10, 50, 100]

# Plotting the binomial distributions and normal approximations
for n in n_values:
    # Binomial distribution
    x = np.arange(0, n+1)
    binomial_pmf = binom.pmf(x, n, p)
    plt.bar(x, binomial_pmf, alpha=0.5, label=f'Binomial n={n}')

    # Normal approximation
    mu = n * p
    sigma = np.sqrt(n * p * (1-p))
    x_norm = np.linspace(0, n, 1000)
    normal_pdf = norm.pdf(x_norm, mu, sigma)
    plt.plot(x_norm, normal_pdf, label=f'Normal Approx n={n}', linewidth=2)

plt.title('Convergence of Binomial to Normal')
plt.legend()
plt.show()
````

## 4. Standard Normal Distribution and it's connection with Z score

The Standard Normal Distribution is a special case of the normal distribution where the mean (\( \mu \)) is 0 and the standard deviation (\( \sigma \)) is 1. The random variable associated with a standard normal distribution is often denoted as \( Z \).

### Z-Score:
A Z-score is a statistical measurement that describes a value's relationship to the mean of a group of values. It is measured in terms of standard deviations from the mean. The formula to calculate a Z-score from a given value \( x \) is:

$$ Z = \frac{{x - \mu}}{{\sigma}} $$

where:
- \( x \) is the value,
- \( \mu \) is the mean of the distribution,
- \( \sigma \) is the standard deviation of the distribution.

The Z-score tells you how many standard deviations a particular value is from the mean. A Z-score of 0 indicates that the data point's score is identical to the mean score, while a Z-score of 1.0 indicates a value that is one standard deviation from the mean.

### Example:
Let's suppose we have a normal distribution with a mean of 50 and a standard deviation of 10. We want to find the Z-score of a value \( x = 70 \).

Using the formula, we get:

$$ Z = \frac{{70 - 50}}{{10}} = 2.0 $$

##  Confidence Intervals

A Confidence Interval (CI) is a range of values, derived from a data set, that is likely to contain the true value of an unknown population parameter. Confidence intervals provide a measure of the uncertainty or margin of error around a point estimate from sample data.

### Key Components:
1. **Point Estimate**: A single value estimated for a population parameter. Common point estimates include the sample mean (\( \bar{x} \)), sample proportion (\( \hat{p} \)), etc.
2. **Margin of Error**: The range within which the true population parameter is likely to fall. It's often denoted by a plus-minus value around the point estimate.
3. **Confidence Level**: The probability that the confidence interval contains the true population parameter. Common confidence levels include 90%, 95%, and 99%.

### Calculation:
The formula to calculate a confidence interval for a population mean (assuming a normal distribution) is:

$$ CI = \bar{x} \pm z \left( \frac{\sigma}{\sqrt{n}} \right) $$

where:
- x bar is the sample mean,
- \( z \) is the z-score (which corresponds to the desired confidence level),
- \( \sigma \) is the population standard deviation,
- \( n \) is the sample size.

### Example:
Let's say an online retailer wants to estimate the average purchase amount from their website. They take a random sample of 100 purchases and find a sample mean of $75 with a standard deviation of $20. They want to estimate the population mean purchase amount with a 95% confidence interval.


````python
import scipy.stats as stats
import numpy as np

# Given data
x_bar = transaction_df.loc[transaction_df['main_event'] == 'Payment Completed','event_value'].mean()  # Sample mean ($)
sigma = transaction_df.loc[transaction_df['main_event'] == 'Payment Completed','event_value'].std()  # Standard deviation ($)
n = 1000  # Sample size
alpha = 0.05  # Significance level (1 - Confidence Level)

# Z-score for a 95% confidence interval
z = stats.norm.ppf(1 - alpha/2)

# Margin of Error
margin_of_error = z * (sigma / np.sqrt(n))

# Confidence Interval
confidence_interval = (x_bar - margin_of_error, x_bar + margin_of_error)

confidence_interval
````

````python
# Given data for conversion rates
n_conversions = 500  # Number of trials (visitors)
x_conversions = 50  # Number of successes (conversions)
p_hat = x_conversions / n_conversions  # Sample conversion rate

# Z-score for a 95% confidence interval
z_conversions = stats.norm.ppf(1 - alpha/2)

# Margin of Error for conversion rates
margin_of_error_conversions = z_conversions * np.sqrt((p_hat * (1 - p_hat)) / n_conversions)

# Confidence Interval for conversion rates
confidence_interval_conversions = (p_hat - margin_of_error_conversions, p_hat + margin_of_error_conversions)

confidence_interval_conversions
````

## One Sample Z-Test in Python

Suppose the IQ in a certain population is normally distributed with a mean of μ = 100 and a standard deviation of σ = 15.

One Sample Z-Test in Python

````python
from statsmodels.stats.weightstats import ztest as ztest
"""
H0 : New Drug not Affects the IQ
H1 : New Drug Does Affects the IQ
"""

#enter IQ levels for 20 patients
data = [88, 92, 94, 94, 96, 97, 97, 97, 99, 99,
        105, 109, 109, 109, 110, 112, 112, 113, 114, 115]

#perform one sample z-test
ztest(data, value=100)
````

In majority of cases we do not have full information about the population especially about the standard deviation. Therefore, T-test is based on the sample std and can be used in this cases. However, if the size of the sample is high Z-test and T-test are about the same.

3. 4. T-test has three main types: One Sample t-test, two-sample t-test (unpaired or independent), and paired t-test.
5. One-sample t-tests compare the sample mean to the hypothesized values, whereas two-sample t-tests (also known as independent or unpaired t-tests) compare the means of two independent groups. The differences between the two dependent variables are compared in a paired t-test.

## Chi-Square Test:

The Chi-Square test is a statistical test used to determine whether there is a significant association between two categorical variables in a sample. It's often used with categorical data to compare observed frequencies with expected frequencies based on certain assumptions.

There are three types of Chi-Square tests:

1. **Chi-Square Goodness of Fit Test**:
   - This tests whether the observed frequencies for a categorical variable match an expected distribution.

2. **Chi-Square Test of Independence**:
   - This tests whether there is a significant association between two categorical variables.

3. **Chi-Square Test of Homogeneity**:
   - This tests whether the distribution of a categorical variable is the same across different groups.

### Example of Chi-Square Test of Independence:

Suppose an online retailer wants to know if there's an association between the region of a customer (North, South, East, West) and the type of product they purchase (Clothing, Electronics, Groceries).

They collect a random sample of purchases and create a contingency table:

```plaintext
            | Clothing | Electronics | Groceries | Row Total
-----------------------------------------------------------
North      |   30     |     40      |    30      |   100
South      |   40     |     30      |    30      |   100
East       |   20     |     50      |    30      |   100
West       |   10     |     30      |    60      |   100
-----------------------------------------------------------
Column Total| 100      |    150      |   150      |   400
```

Now they want to test if the region and type of product are independent.



````python
import scipy.stats as stats
import numpy as np

# Observed data in a (4x3) contingency table
observed = np.array([[30, 40, 30],
                     [40, 30, 30],
                     [20, 50, 30],
                     [10, 30, 60]])

# Perform Chi-Square Test of Independence
chi2_stat, p_value, dof, ex = stats.chi2_contingency(observed)

chi2_stat, p_value
````

## ANOVA
ANOVA (Analysis of Variance) is a statistical method used to test differences between two or more means. It's useful for comparing the means of multiple groups and understanding if the differences between them are statistically significant.

There are several types of ANOVA, including:

1. **One-Way ANOVA**: Tests whether there are significant differences between the means of two or more independent groups.
2. **Two-Way ANOVA**: Tests whether there are significant differences between the means of two or more groups while considering two factors.
3. **Repeated Measures ANOVA**: Tests whether there are significant differences between the means of two or more groups when the same subjects are measured under each condition.

### One-Way ANOVA Example:

Suppose an online retailer wants to compare the average purchase amounts between four different types of marketing campaigns. They collect the following data on purchase amounts from a random sample of customers:

```plaintext
Campaign A: $50, $60, $70, $80, $90
Campaign B: $30, $40, $50, $60, $70
Campaign C: $20, $30, $40, $50, $60
Campaign D: $10, $20, $30, $40, $50
```

They want to test if there is a significant difference in the average purchase amount between these campaigns.


````python
import scipy.stats as stats
import numpy as np

# Data
data_a = np.array([50, 60, 70, 80, 90])
data_b = np.array([30, 40, 50, 60, 70])
data_c = np.array([20, 30, 40, 50, 60])
data_d = np.array([10, 20, 30, 40, 50])

# Perform One-Way ANOVA
f_stat, p_value = stats.f_oneway(data_a, data_b, data_c, data_d)

f_stat, p_value
````


The p-value will tell you whether there are significant differences between the means of the campaigns. If the p-value is less than the significance level (e.g., 0.05), you reject the null hypothesis and conclude that there's a significant difference between at least two of the campaigns.

## Task 1

### Description:
In this task, the goal is to test whether the conversion rates of different variants are significantly different by more than 10% relatively. This is an example of a hypothesis test where we will compare the proportions of successes (conversions) between different groups (variants) to see if the differences are statistically significant.

### Hypothesis:
- **Null Hypothesis (\(H_0\))**: There is no difference in conversion rates between the variants, or the relative difference is equal to 10%.
- **Alternative Hypothesis (\(H_1\))**: The relative difference in conversion rates between the variants is more than 10%.

### Procedure:
1. **Prepare the Data**:
   - Ensure the data is correctly loaded and structured, with columns for variants and conversions.
   
2. **Calculate the Conversion Rates**:
   - Calculate the conversion rates for each variant.

3. **Perform the Hypothesis Test**:
   - Use a statistical test to compare the conversion rates of different variants.
   
4. **Interpret the Results**:
   - Based on the p-value from the hypothesis test, decide whether to reject the null hypothesis in favor of the alternative hypothesis.

### Solution:

- We first calculate the conversion rates for each variant using `groupby` and `mean`.
- We then calculate the observed relative difference in conversion rates between the two variants.
- We perform a two-proportion z-test using `stats.proportions_ztest` to test for a 10% relative difference in conversion rates.
- Finally, we output the observed relative difference, z-statistic, and p-value from the hypothesis test.


````python
import statsmodels.stats.proportion as smp
import pandas as pd

# Assuming data is loaded and structured correctly
# Conversion rates for each variant
conversion_rates = data.groupby('variant')['conversion'].mean()

# Assuming 'A' and 'B' are the names of the variants
# Calculate the observed relative difference
observed_relative_difference = abs(conversion_rates['A'] - conversion_rates['B']) / conversion_rates['A']

# Perform hypothesis test
# For simplicity, assume we're comparing two variants: A and B
# Prepare data for the two-proportion z-test
successes = data.groupby('variant')['conversion'].sum()
nobs = data['variant'].value_counts()

# Two-proportion z-test
z_stat, p_value = smp.proportions_ztest(successes, nobs, value=0.1, alternative="larger")  # testing for a 10% relative difference

observed_relative_difference, z_stat, p_value
````

## Task 2

### Task Description:
In this task, we aim to test whether the revenue per user (RPU) of different variants differs by more than 10% relatively. This involves a hypothesis test where we compare the means of revenue per user between different groups (variants) to assess if the differences observed are statistically significant.

### Hypotheses:
- **Null Hypothesis (\(H_0\))**: The relative difference in revenue per user between the variants is equal to 10%.
- **Alternative Hypothesis (\(H_1\))**: The relative difference in revenue per user between the variants is more than 10%.

### Procedure:
1. **Prepare the Data**:
   - Ensure the data is correctly loaded and structured, with columns for variants and revenue.

2. **Calculate the Revenue Per User (RPU)**:
   - Calculate the mean revenue per user for each variant.

3. **Perform the Hypothesis Test**:
   - Use a statistical test to compare the revenue per user of different variants.
   
4. **Interpret the Results**:
   - Based on the p-value from the hypothesis test, decide whether to reject the null hypothesis in favor of the alternative hypothesis.

### Solution:
- We first calculate the Revenue Per User (RPU) for each variant using `groupby` and `mean`.
- We then calculate the observed relative difference in RPU between the two variants.
- We perform an independent two-sample t-test using `stats.ttest_ind` to compare the means of revenue per user between the two variants.
- Finally, we output the observed relative difference, t-statistic, and p-value from the hypothesis test.

````python
import scipy.stats as stats

# Assuming data is loaded and structured correctly
# Revenue Per User (RPU) for each variant
rpu = data.groupby('variant')['revenue'].mean()

# Assuming 'A' and 'B' are the names of the variants
# Calculate the observed relative difference
observed_relative_difference_rpu = abs(rpu['A'] - rpu['B']) / rpu['A']

# Perform hypothesis test
# We'll use an independent two-sample t-test to compare the means of revenue per user between the two variants
# Prepare data for the t-test
revenue_a = data[data['variant'] == 'A']['revenue']
revenue_b = data[data['variant'] == 'B']['revenue']

# Two-sample t-test
t_stat_rpu, p_value_rpu = stats.ttest_ind(revenue_a, revenue_b, equal_var=False)  # Assuming unequal variances

observed_relative_difference_rpu, t_stat_rpu, p_value_rpu
````
